```python
import pandas as pd
df = pd.read_csv('/content/07_scheme_performance.csv')
df.head()
```

### Cleaning `07_scheme_performance.csv`

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

```python
df['amfi_code'] = df['amfi_code'].astype(str)
```

```python
text_cols = [
    'scheme_name',
    'fund_house',
    'category',
    'plan',
    'risk_grade'
]

for col in text_cols:
    df[col] = (
        df[col]
        .astype(str)
        .str.strip()
        .str.title()
    )
```

```python
numeric_cols = [
    'return_1yr_pct',
    'return_3yr_pct',
    'return_5yr_pct',
    'benchmark_3yr_pct',
    'alpha',
    'beta',
    'sharpe_ratio',
    'sortino_ratio',
    'std_dev_ann_pct',
    'max_drawdown_pct',
    'aum_crore',
    'expense_ratio_pct',
    'morningstar_rating'
]

df[numeric_cols] = df[numeric_cols].apply(
    pd.to_numeric,
    errors='coerce'
)
```

```python
return_cols = [
    'return_1yr_pct',
    'return_3yr_pct',
    'return_5yr_pct',
    'benchmark_3yr_pct'
]


for col in return_cols:
    print(
        col,
        "Non Numeric Values:",
        df[col].isnull().sum()
    )
```

```python
df['anomaly_flag'] = False
```

```python
df.loc[
    (df['return_1yr_pct'] > 100) |
    (df['return_1yr_pct'] < -100),
    'anomaly_flag'
] = True
```

```python
df.loc[
    (df['expense_ratio_pct'] < 0.1) |
    (df['expense_ratio_pct'] > 2.5),
    'anomaly_flag'
] = True
```

```python
df.loc[
    (df['sharpe_ratio'] > 10) |
    (df['sharpe_ratio'] < -10),
    'anomaly_flag'
] = True
```

```python
print(
    "Total Anomalies:",
    df['anomaly_flag'].sum()
)
```

```python
print(
    "Invalid Expense Ratio:",
    (
        (df['expense_ratio_pct'] < 0.1) |
        (df['expense_ratio_pct'] > 2.5)
    ).sum()
)
```

```python
print('Missing values before fillna:')
print(df[numeric_cols].isnull().sum())

df[numeric_cols] = df[numeric_cols].fillna(
    df[numeric_cols].median()
)

print('\nMissing values after fillna:')
print(df[numeric_cols].isnull().sum())
```

```python
df[text_cols] = df[text_cols].fillna("Unknown")
```

```python
print("AUM less than 0:", (df['aum_crore'] < 0).sum())
df = df[df['aum_crore'] >= 0]
```

```python
print("Expense ratio less than 0:", (df['expense_ratio_pct'] < 0).sum())
df = df[df['expense_ratio_pct'] >= 0]
```

```python
print("Invalid Morningstar ratings:", df[~df['morningstar_rating'].isin([1,2,3,4,5])].shape[0])
df = df[
    (df['morningstar_rating'] >= 1) &
    (df['morningstar_rating'] <= 5)
]
```

```python
print("Beta less than 0:", (df['beta'] < 0).sum())
df = df[df['beta'] >= 0]
```

```python
df = df.sort_values(
    ['fund_house','category']
).reset_index(
    drop=True,
    inplace=False
)
```

```python
print("Final Dataset Shape:", df.shape)

print("\nExpense Ratio Range:")
print(df['expense_ratio_pct'].min())
print(df['expense_ratio_pct'].max())


print("\nAnomaly Count:")
print(df['anomaly_flag'].value_counts())
```

```python
import os

output_dir = '../Cleaned_Data'
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

df.to_csv(os.path.join(output_dir, 'Cleaned_07_scheme_performance.csv'), index=False)
```

```python
df = df.drop_duplicates()

df.shape
```

```python
df.isnull().sum()
```

```python
num_cols = df.select_dtypes(include=['int64','float64']).columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
df['transaction_type'].unique()
```

```python
df['transaction_type'] = (
    df['transaction_type']
    .str.strip()
    .str.lower()
)

df['transaction_type'] = df['transaction_type'].replace({
    'sip':'SIP',
    'lumpsum':'Lumpsum',
    'lump sum':'Lumpsum',
    'lump-sum':'Lumpsum',
    'redemption':'Redemption'
})
```

```python
print(df['transaction_type'].value_counts())
```

```python
print("Invalid Amount:",
      (df['amount_inr'] <= 0).sum())
```

```python
df = df[df['amount_inr'] > 0]
```

```python
df['transaction_date'] = pd.to_datetime(
    df['transaction_date'],
    errors='coerce'
)
```

```python
print("Invalid Dates:",
      df['transaction_date'].isnull().sum())
```

```python
df = df.dropna(subset=['transaction_date'])
```

```python
df['kyc_status'].unique()
```

```python
valid_kyc = [
    'Verified',
    'Pending',
    'Rejected'
]


invalid_kyc = df[
    ~df['kyc_status'].isin(valid_kyc)
]


print("Invalid KYC Records:",
      invalid_kyc.shape[0])
```

```python
df = df[df['kyc_status'].isin(valid_kyc)]
```
